### Лабораторная работа №2

В файле `languages.csv` сохранены данные по вымирающим/вымершим языкам и языкам на грани исчезновения. Одна строка таблицы соответствует одному языку. 

Показатели в файле:

* `name_en`: название языка на английском;
* `countries`: страны, на которых говорят на данном языке;
* `status`: статус языка – степень угрозы существованию языка согласно ЮНЕСКО, см. ниже;
* `speakers`: число говорящих на языке;
* `location`: описание ареала языка (где распространен);
* `lat` и `lon` : широта и долгота для `location`.

Пояснения для значений в `status`:

* 1: `Vulnerable`, большинство детей говорят на языке, но чаще всего в ограниченных сферах деятельности (например, дома); 
* 2: `Definitely endangered`, дети больше не изучают язык как родной дома; 
* 3: `Severely endangered`, на языке говорит старшее поколение (бабушки/дедушки), родители могут язык понимать, но не говорить на нем со своими детьми; 
* 4: `Critically endangered`, самые молодые носители языка - старшее поколение (бабушки/дедушки и старше), но они говорят на языке редко;
* 5: `Extinct`, говорящих на языке не осталось.

Ниже написан код, который загружает данные из файла по ссылке, забирает из датафрейма столбцы с названиями `status` и `speakers`, а из столбцов – массивы значений.

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyAll26/refs/heads/main/languages.csv")

speakers = df["speakers"].values
status = df["status"].values

print(speakers)
print(status)
```

### Задача 8

Сохраните в массив `typical` значения массива `speakers`, которые являются **типичными** по критерию 1.5 $\times$ IQR. Вычислите число **нетипичных** значений с точки зрения этого критерия.

```python
# считаем границы по критерию 1.5 * IQR
# [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR]

Q1 = np.quantile(speakers, 0.25)
Q3 = np.quantile(speakers, 0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

typical = speakers[(speakers >= lower) & (speakers <= upper)]

# чтобы не отфильтровывать еше и нетипичные значения, 
# сравним длину исходного массива и длину массива typical,
# это и будет число нетипичных значений

print(len(speakers) - len(typical))
```

### Задача 9

Сохраните в массив `non_typical` значения массива `speakers`, которые являются **нетипичными** по критерию 2 $\times$ STD. Вычислите, сколько наблюдений будет в массиве `speakers`, если мы оставим в нем только значения, которые считаются типичными с точки зрения этого критерия.

```python
# считаем границы по критерию 2 * STD
# [mean - 2 * STD, mean + 2 * STD]

MEAN = speakers.mean()
STD = speakers.std(ddof = 1)

low = MEAN - 2 * STD
upp = MEAN + 2 * STD

non_typical = speakers[(speakers < low) | (speakers > upp)]

# чтобы не отфильтровывать еше и типичные значения, 
# сравним длину исходного массива и длину массива non_typical,
# это и будет число типичных значений

print(len(speakers) - len(non_typical))
```
