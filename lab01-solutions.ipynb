### Задача 1

В списке `words` сохранены слова. Используя функцию `map()`, создайте новый список, состоящий из количества букв в каждом слове.

### Задача 5

* `coffee_price`: цены разных видов кофе;

### Задача 6

В массивах сохранены данные по заказам разных видов кофе в кофейне «Случайный лес», сделанным 10 посетителями:

* `name`: названия заказанных видов кофе;
* `price`: цены разных видов кофе;
* `milk`: факт приготовления кофе на альтернативном молоке (1 – альтернативное, 0 – обычное), добавляет 40 рублей к стоимости.

Известно, что в период наблюдения в кофейне действует скидка 10% на заказ любого вида эспрессо.

Создайте массив с *итоговой стоимостью* заказа посетителей – сколько заплатил с учетом всех факторов каждый из 10 гостей.

```python
name = np.array(["Эспрессо Дикий бутстрап", 
                 "Латте Решающий пень в весенний день", 
                 "Эспрессо Дикий бутстрап",
                 "Капучино Священный рандом", 
                 "Капучино Священный рандом", 
                 "Латте Решающий пень в весенний день", 
                 "Эспрессо Утро в случайном лесу", 
                 "Латте Дедлайн, который превратился в тыкву", 
                 "Фильтр-кофе Простокофе", 
                 "Капучино Священный рандом"])

price = np.array([220, 300, 220, 270, 270, 300, 220, 300, 180, 270])
milk = np.array([0, 0, 0, 1, 0, 1, 0, 0, 0, 1])

# массив из 0 и 1 для молока уже есть ->
# массив milk * 40 – сколько каждый заплатит за молоко, 0 * 40 = 0 или 1 * 40 = 40 руб

# по тому же принципу создаем массив esp из 0 и 1 (False и True), где 1 – Эспрессо разных видов
# массив esp * price * 0.4 – сколько каждому скинут от стоимости заказа, 
# 0, если в esp 0 (не эспрессо) или 40% от price, если в esp 1 (эспрессо)

esp = (name == "Эспрессо Дикий бутстрап") | (name == "Эспрессо Утро в случайном лесу")

final = price + milk * 40 - esp * price * 0.4
print(final)
```

### Задача 7* (дополнительно на векторизацию)

Напишите функцию `convert()`, которая принимает на вход слово с названием цвета (считаем, что их два, `black` и `white`) и «производит» инверсию – черный цвет меняет на белый, и наоборот. Другими словами, на слово `black` функция должна возвращать `white`, а на слово `white` – `black`.

Векторизуйте эту функцию и получите на основе списка `colors` список инвертированных цветов.

```python
colors = ["black", "white", "white", "black", "black", "white", 
         "white", "white", "black", "black", "black"]

def convert(x):
    if x == "black":
        y = "white"
    else:
        y = "black"
    return y

conv = np.vectorize(convert)

inv_colors = conv(colors)
print(inv_colors)
```
